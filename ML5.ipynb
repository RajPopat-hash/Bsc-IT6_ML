```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/01 covid_toy.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,0:5]
y=df.drop(x,axis='columns')
print (x.head())
print()
print()
print()
print(y.head())
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
print(x_train.shape)
print()
print()
print(y_train.shape)
#
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
si=SimpleImputer()

le=LabelEncoder()
```

```python
y_train.has_covid=le.fit_transform(y_train.has_covid)
y_test.has_covid=le.fit_transform(y_test['has_covid'])
x_train['fever']=si.fit_transform(x_train[['fever']])
x_test['fever']=si.fit_transform(x_test[['fever']])
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop='first',sparse_output=False)
x_train_gender_city=ohe.fit_transform(x_train[['gender','cough']])
x_test_gender_city=ohe.fit_transform(x_test[['gender','cough']])
print(x_train_gender_city.shape)
print(x_test_gender_city.shape)
```

```python
from sklearn.compose import ColumnTransformer

transformer=ColumnTransformer(transformers=[
    ('tf1',SimpleImputer(),['fever']),
    ('tf2',OneHotEncoder(drop='first',sparse_output=False),['gender','cough'])
],remainder='passthrough')

# Using the ColumnTransformer to fit and transform the datasets automatically
x_train_transformed = transformer.fit_transform(x_train)
x_test_transformed = transformer.transform(x_test)

print("x_train_transformed shape:", x_train_transformed.shape)
print("x_test_transformed shape:", x_test_transformed.shape)
```
